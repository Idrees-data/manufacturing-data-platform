```python
machine_id = [f"M00{x}" for x in range(100)]
machine_name = [f"CNS-0{x}" for x in range(100)]
import random
factory_place = ["Hyderabad","Mumbai","Goa","Banglore","Karnatak"]
factory = [random.choice(factory_place) for _ in range(100)]
import pandas as pd
import numpy as np

# Define range and number of dates
start_date = pd.to_datetime('2026-01-01')
end_date = pd.to_datetime('2026-12-31')
num_dates = 100

# Generate installation_date using unix timestamps
start_u = start_date.value // 10**9
end_u = end_date.value // 10**9
random_timestamps = np.random.randint(start_u, end_u, num_dates)
installation_date = pd.to_datetime(random_timestamps, unit='s')

#shift Morning
shifs = ["Morning","Nights"]
shifts = [random.choice(shifs) for x in range(100)]

# production_line 
production_l = ["L1","L2",'L3',"L4"]
production_line = [ random.choice(production_l) for x in range(100)]

#status
state = ["Active","Inactive"]
status = [random.choice(state) for x in range(100)]
```

```python
machinary = pd.DataFrame({
    "machine_id":machine_id,
    "machine_name":machine_name,
    "factory":factory,
    "installation_date":installation_date,
    "shifts":shifts,
    "production_line":production_line,
    "status":status
})
machinary
```

```python
machinary.to_csv("Machines.csv", index= False)
```

```python

maintenance_id = [f"MT000{x}" for x in range(100)] #MT0001
machine_id
maintenance_date = [ d + pd.Timedelta(days= random.randint(1,100))
                    for d in installation_date]
maintenance_date

maintenance_t = ["Preventive","Corrective","Predictive"]
maintenance_type = [random.choice(maintenance_t) for _ in range(100)]
maintenance_type


downtime_rules = {
    "Preventive": (1, 4),
    "Predictive": (1, 3),
    "Corrective": (3, 12)
}
downtime_hours = [
    round(random.uniform(downtime_rules[b][0],downtime_rules[b][1]),1)
    for b in maintenance_type
]

maintenance_ = {
        "Preventive": (200, 800),
        "Predictive": (3000, 10000),
        "Corrective": (8000, 30000)
}

maintenance_cost = [
    round(random.uniform(maintenance_[m][0],maintenance_[m][1]),1)
    for m in maintenance_type
]
```

```python
maintenance = pd.DataFrame({
    'maintenance_id':maintenance_id,
    'machine_id':machine_id,
    'maintenance_date':maintenance_date,
    'maintenance_type':maintenance_type,
    'downtime_hours':downtime_hours,
    'maintenance_cost':maintenance_cost
})

maintenance.to_csv('Maintenance.csv',index= False)
```

```python
inspection_id = [f"Q000{x}" for x in range(100)] #Q0001
production_id = [f"PR000{x}" for x in range(100)] #PR0001
inspection_date = [
    d + pd.Timedelta(days=random.randint(0, 2))
    for d in production_date
]

defect_ty = ['Surface Defect','Dimensional Error','Assembly Error']
defect_type = [random.choice(defect_ty) for x in range(100)]
```

```python
units_produced = [
    random.randint(50, 200)
    for _ in range(100)
]

inspected_units = [
    random.randint(int(u * 0.7), u)
    for u in units_produced
]

defective_units = [
    random.randint(0, int(i * 0.10))
    for i in inspected_units
]
```

```python
Quality_Inspections = pd.DataFrame({
    'inspection_id':inspection_id,
    'production_id':production_id,
    'inspection_date':inspection_date,
    'inspected_units':inspected_units,
    'inspected_units':inspected_units,
    'inspected_units':inspected_units,
    'inspected_units':inspected_units,
    'inspected_units':inspected_units,
    'defective_units':defective_units,
    'defective_units':defective_units,
    'defective_units':defective_units,
    'defect_type':defect_type
})

Quality_Inspections
```

```python
Quality_Inspections.to_csv('Quality_Inspections.csv',index = False)
```
